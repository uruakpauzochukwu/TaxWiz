```python
class Chunk:
    def __init__(self, text, metadata):
        self.text = text
        self.metadata = metadata
```

```python
import os
import pickle
from pathlib import Path

# Current notebook folder
notebook_dir = Path().resolve()  # e.g., .../TaxWiz/notebooks
project_root = notebook_dir.parent  # move up one level to TaxWiz/

# Build path relative to project root
chunks_file = project_root / "chunks_dict.pkl"

# Load pickle
with open(chunks_file, "rb") as f:
    extracted_text = pickle.load(f)

print("File loaded successfully")
```

```python
# Correctly extract all text chunks
all_chunks = []

for item in extracted_text:
    # Extract the string under the 'content' key
    text = item.get("content", "")
    
    # Skip empty strings
    if text.strip():
        all_chunks += chunk_text(text=text, chunk_size=1050, overlap=50)

print(f"Total chunks: {len(all_chunks)}")
```
